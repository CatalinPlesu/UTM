# 1. Tokenisation

```python
from nltk.tokenize import word_tokenize

text_column = data['text']
cloumn_tokens = []

for row in text_column:
    cloumn_tokens.append(word_tokenize(row))

print(cloumn_tokens[:5])
```

# 2.  Finding frequency distinct in the text

```python
from nltk.probability import FreqDist

fdists = []
for token in cloumn_tokens:
    fdists.append(FreqDist(token))

fdist1s = []
for fdist in fdists:
    fdist1s.append(fdist.most_common())

print(fdist1s[:5])
```

```python
all_tockens = []
for token in cloumn_tokens:
    all_tockens.extend(token)

fdist1 = FreqDist(all_tockens).most_common()
print(fdist1[:1000])
```

# 3. Stemming

```python
from nltk.stem import PorterStemmer

pst = PorterStemmer()
print(len(set(all_tockens)))

stemmed = []
for unique in set(all_tockens):
    stemmed.append(pst.stem(unique))

print(len(set(stemmed)))
# print(set(stemmed))
print(list(set(stemmed))[:50])
```

# 4. Lemmatisation

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

lematised = []
for unique in set(all_tockens):
    lematised.append(lemmatizer.lemmatize(unique))

print(len(set(lematised)))
print(list(set(lematised))[:50])
```

# 5. Stop Words

```python
from nltk.corpus import stopwords
a = set(stopwords.words("english"))

# text = "Cristiano Ronaldo was born on February 5, 1985, in Funchal, Madeira, Portugal."
# text1 = word_tokenize(text.lower())
# print(text1)

without_stopwords = [x for x in set(stemmed) if x not in a]
print(len(without_stopwords))
print(without_stopwords[:1000])
```

# 6. Part of speech tagging (POS)

```python
i = 0
for tex in cloumn_tokens:
    i += 1
    print("############")
    print(tex)
    for token in tex:
        print(nltk.pos_tag([token]))
    if i == 5:
        break
```

# 7. Named entity recognition

```python
from nltk import ne_chunk

i = 0
for token in cloumn_tokens:
    i += 1
    print("############")
    tags = nltk.pos_tag(token)
    chunk = ne_chunk(tags)
    print(chunk)
    if i == 5:
        break
```

# 8. Chunking

```python
i = 0
for token in cloumn_tokens:
    i += 1
    print("############")

    tags = nltk.pos_tag(token)
    reg = "NP: {<DT>?<JJ>*<NN>}"
    a = nltk.RegexpParser(reg)
    result = a.parse(tags)
    print(result)
    
    if i == 5:
        break
```
